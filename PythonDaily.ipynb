```python
#Set Operations


#1.Union (|) -- Unique elements from both sets
set1={1,2,3,3,4,5,5,6,56}
set2={2,23,43,56,87}

res=set1 | set2
#res=set1.union(set2)
print("Union:",res)

#2.Intersection (&) -- Common elements from both sets
#res= set1 & set2
res=set1.intersection(set2)
print("Intersection:",res)

#3.Difference (-) --  Elements from set1 which are not in set2
res=set1 - set2
#res=set1.difference(set2)
print("Difference:",res)

#4.symmetric diffr () - common ones are removed
res=set1 ^ set2
#res=set1.symmetric_difference(set2)
print("Symmetric Difference:",res)

#5.SubSet
set3={88,99}
set4={1,2,88,99}
res=set3.issubset(set2)
print("SubSet:",res)

#6.SuperSet
res=set4.issuperset(set3)
print("SuperSet:",res)

#7.Disjoint - when 2 have no common elements
set3={88,99}
set4={1,2,3,4}
res=set3.isdisjoint(set4)
print("Disjoint:",res)
```
